```python
import os
import wfdb
import numpy as np
```

```python
AFDB_PATH = "/home/intellisense01/EML-Labs/datasets/afdb"
os.listdir(AFDB_PATH)
```

```python
datafiles = [f.split('.')[0] for f in os.listdir(AFDB_PATH) if (f.endswith('.qrs') or f.endswith('.qrsc'))]
datafiles
```

```python
print(len(datafiles))
```

```python
probalamatic_files = [
    "00735",
    "03665",
    "04043",
    "04936",
    "05091",
    "06453",
    "08378",
    "08405",
    "05091",
    "06453",
    "08378",
    "08405",
    "08434",
    "08455",
]
```

```python
filtered_datafiles = [f for f in datafiles if f not in probalamatic_files]
```

```python
print(len(filtered_datafiles))
```

```python
af_patients = []
for file in filtered_datafiles:
    file_path = os.path.join(AFDB_PATH,file)
    ann = wfdb.rdann(file_path, "atr")
    print(np.unique(ann.aux_note))
    if '(AFIB' in np.unique(ann.aux_note) and '(N' in np.unique(ann.aux_note) and len(np.unique(ann.aux_note)) == 2:
        af_patients.append(file)
print(af_patients)
```
